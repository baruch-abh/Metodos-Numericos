**ITERACION DE PUNTO FIJO**

El metodo de punto fijo es un algoritmo numerico diseñado para encontrar las raices de yuna ecuacion $$f(x)=0$$ transformandola a la forma $$x=g(x)$$. se comienza con $$p_0$$ y utiliza la formula iterativa $$p_{n+1}=g(p_n)$$ para generar una sucesion de valores. si el despeje $$g(x)$$ estas aproximaciones convergen a la solucion exacta.


En este codigo se muestra la solucion de la funcion $$f(x)=x^3+4x^2-10$$, para lograrlo se utilizo el despeje $$x=x-\frac{f(x)}{f´(x)}$$ es decir $$x=x-\frac{x^3+4x^2-10}{3x^2+8x}$$

este programa ejecuta un ciclo con un numero predefinido de iteraciones donde calcula las aproximaciones de $$x$$ y almacena el error absoluto, asi se puede ver conforme al progreso como disminuye el factor de error
