Metodo Biseccion

queremos resolver la ecuacion $x^3+4x^2-10=0$ en el intervalo $[1,2]$

Para aplicar el **metodo de biseccion** verificamos que la funcion $f(x)$ evaluada en los extremos opuestos del intervalo $[a,b]$ tiene signos opuestos, es decir, $f(a) \times f(a)<0$.

El teorema del valor intermedio nos asegura que existe un numero $p \in [a,b] $ tal que es una raiz de $f(x)$, es decir, $f(p)=0$

Aproximaremos la solucion, calculando de manera iterativa el punto medio del intervalo: $$p_i=\frac{(b_i-a_i)}2$$




importamos  la libreria numpy, que tiene funciones matematicas y matplot lib, que permite graficas

```python
import numpy as np
#ingresamos datos de entrada para los diferentes metodos a trabajar

a = 1
b = 2

#guarda valores iniciales

a0=a
b0=b

tol = 0.000001
nmax = 100
error = 100
niter = 0
def f(x):
  return x**3 + 4*x**2 - 10.0
# Método de Bisección

# evaluo primer valor medio
m = a + (b - a)/2

#Evaluacion de la función en los puntos a, b y m
fa = f(a)
fb = f(b)
fm = f(m)

print("# iter\t\t a \t\t f(a) \t\t b \t\t f(b) \t\t m \t\t f(m) \t\t error")
print("{0} \t\t {1:6.4f} \t {2:6.4f} \t {3:6.4f} \t {4:6.4f} \t {5:6.4f} \t {6:6.4f} \t {7:6.4f}".format(niter, a0, fa, b0, fb, m, fm, error ))

# ciclo iterativo
while error > tol and niter < nmax:
    m = a + (b - a) / 2
    if np.sign(fa) == np.sign(fm):
        a = m
        fa = f(a)
    else:
        b = m
        fb = f(b)

    m = a + (b - a)/2
    fm = f(m)
    error = abs(b - a)
    niter += 1
    print("{0} \t\t {1:6.6f} \t {2:6.6f} \t {3:6.6f} \t {4:6.6f} \t {5:6.6f} \t {6:6.6f} \t {7:6.6f}".format(niter, a, fa, b, fb, m, fm, error ))

print("La raíz de la función dada en el intervalo [{0:6.4f},{1:6.4f}] es {2:6.7f}".format(a0,b0,m))
```
