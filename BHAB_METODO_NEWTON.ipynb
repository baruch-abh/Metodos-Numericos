**APLICACION DE METODO DE NEWTON**

El metodo de punto fijo es un algoritmo numerico diseñado para encontrar las raices de una ecuacion $$f(x)=0$$ transformandola a la forma $$x=g(x)$$. se comienza con $$p_0$$ y utiliza la formula iterativa $$p_{n+1}=g(p_n)$$ para generar una sucesion de valores. si el despeje $$g(x)$$ estas aproximaciones convergen a la solucion exacta.


En este codigo se muestra la solucion de la funcion $$f(x)=cosx -x$$,

para lograrlo de una manera mas rapido y con menos iteraciones se utilizo el despeje del metodo de newton$$x=x-\frac{f(x)}{f´(x)}$$ es decir $$x=x-\frac{cosx-x}{-senx-1}$$
